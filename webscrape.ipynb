We use json5 instead of json as JS objects contain commas which JSON Parsing does not accept. JSON5 supports this.

```python
import json5
```

```python
url='https://www.chess.ca/ext/cfc-data.9e681e5168e3a3385af234d940059e13e7e29cfd40c6dd8ddba056062a238083.js'
```

```python
response = requests.get(url)
response.raise_for_status()
text = response.text
```

raise_For_status checks if HTTP request was successful or not.

```python
start = text.find("{")
end = text.rfind("}") + 1
data = json5.loads(text[start:end])
```

Used to get the start and end of the file with the data so that we can extract it.

```python
import pandas as pd

events = data["events"]

df = pd.DataFrame(events)

print(df.head())

df.to_csv("chess_tournaments.csv", index=False)
```

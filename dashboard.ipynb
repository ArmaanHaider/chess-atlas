```python
import sqlite3
import pandas as pd
import streamlit as st
```

Sqlite is easy to integrate with Python so it is used.

```python
df = pd.read_csv("chess_tournaments.csv")
conn = sqlite3.connect("chess_tournaments.db")
df.to_sql("tournaments", conn, if_exists="replace", index=False)
conn.close()
```
